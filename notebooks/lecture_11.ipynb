$$u(\$10{,}000) = 1, \quad u(\$0) = 0$$

```python
# Use the following format: (Money value, Utility value)
best_outcome = (10000,1)
worst_outcome = (0,0)
```

Next we will investigate multiple points between \$0 and \$10,000 and assign each point an indifference probability $P^*$.

```python
# Add points to the list below using this format:
# points = [(X1, P*),
#            (X2, P*),
#            ...]

points = [ (2000, 0.3),
           (4000, 0.55),
           (6000, 0.8),
           (8000, 0.95)
]
```

Using our selected $P^*$ values for each point, our utility values can be found by solving the following equation:

$$E[u \mid a_1] = E[u \mid a_2]$$

$$u(\text{most favorable outcome}) \cdot P^* + u(\text{least favorable outcome}) \cdot (1 - P^*) = u(X)$$

The cell below contains a function that will calculate and return your utility values

```python
# Run this cell to find your utility function
myutility = utility(best_outcome, worst_outcome, points)
```
